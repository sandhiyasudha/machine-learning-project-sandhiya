```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('placement.csv')

# Display first 5 rows
print(df.head())

# Select feature (X) and target (Y)
# Replace 'cgpa' and 'package' with your actual column names

X = df['cgpa'].values
Y = df['package'].values

# Linear Regression using Ordinary Least Squares
def linear_regression_ols(X, y):

    # Add bias/intercept column
    X_b = np.c_[np.ones((len(X), 1)), X]

    # Calculate theta using OLS formula
    theta = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)

    return theta

# Prediction function
def predict(X, theta):

    # Add bias/intercept column
    X_b = np.c_[np.ones((len(X), 1)), X]

    # Predict values
    y_pred = X_b.dot(theta)

    return y_pred

# Train model
theta = linear_regression_ols(X, Y)

print("Intercept:", theta[0])
print("Slope:", theta[1])

# Predict for new values
X_new = np.array([7.5, 8.0, 8.5])

predictions = predict(X_new, theta)

print("\nPredictions:")
for x, y in zip(X_new, predictions):
    print(f"CGPA = {x} --> Predicted Package = {y:.2f}")

# Plot data points
plt.scatter(X, Y, color='blue', label='Actual Data')

# Regression line
X_line = np.linspace(min(X), max(X), 100)
Y_line = predict(X_line, theta)

plt.plot(X_line, Y_line, color='red', label='Regression Line')

plt.xlabel('CGPA')
plt.ylabel('Package')
plt.title('Linear Regression using OLS')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('placement.csv')

# Remove unnecessary column
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)

# Feature and Target
X = df['cgpa'].values
Y = df['iq'].values

# Linear Regression (OLS)
X_b = np.c_[np.ones((len(X), 1)), X]
theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ Y

intercept = theta[0]
slope = theta[1]

print("Intercept =", intercept)
print("Slope =", slope)

# Predicted values for all data points
Y_pred = intercept + slope * X

# Sort for smooth line
sorted_idx = np.argsort(X)
X_sorted = X[sorted_idx]
Y_pred_sorted = Y_pred[sorted_idx]

# Plot
plt.figure(figsize=(10, 6))

# Actual dataset points
plt.scatter(X, Y, s=60, label='Actual Data')

# Regression line through all data
plt.plot(X_sorted, Y_pred_sorted,
         linewidth=3,
         label='Regression Line')

plt.xlabel('CGPA')
plt.ylabel('IQ')
plt.title('Linear Regression: CGPA vs IQ')
plt.grid(True)
plt.legend()

plt.show()

# Predict new values
new_cgpa = np.array([6.0, 7.0, 8.0])
new_predictions = intercept + slope * new_cgpa

print("\nPredictions:")
for cgpa, iq in zip(new_cgpa, new_predictions):
    print(f"CGPA = {cgpa} --> Predicted IQ = {iq:.2f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('placement.csv')

# Remove unwanted column if present
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)

# Display dataset information
print("First 5 Records:")
print(df.head())

print("\nColumns:")
print(df.columns)

# Feature (X) and Target (Y)
X = df['cgpa'].values
Y = df['iq'].values

# Linear Regression using OLS
def linear_regression_ols(X, Y):
    X_b = np.c_[np.ones((len(X), 1)), X]
    theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ Y
    return theta

# Prediction function
def predict(X, theta):
    X_b = np.c_[np.ones((len(X), 1)), X]
    return X_b @ theta

# Train model
theta = linear_regression_ols(X, Y)

print("\nModel Parameters")
print("Intercept =", theta[0])
print("Slope =", theta[1])

# Predictions for new values
X_new = np.array([6.0, 7.0, 8.0])
predictions = predict(X_new, theta)

print("\nPredictions")
for x, y in zip(X_new, predictions):
    print(f"CGPA = {x} --> Predicted IQ = {y:.2f}")

# Create regression line
X_line = np.linspace(0, max(X) + 1, 100)
Y_line = predict(X_line, theta)

# Plot graph
plt.figure(figsize=(10, 6))

# Actual data points
plt.scatter(X, Y, color='blue', s=50, label='Actual Data')

# Regression line
plt.plot(X_line, Y_line,
         color='red',
         linewidth=3,
         label='Regression Line')

# Force graph to start from 0
plt.xlim(0, 10)
plt.ylim(0, max(Y) + 20)

plt.xlabel('CGPA')
plt.ylabel('IQ')
plt.title('Linear Regression using OLS (CGPA vs IQ)')
plt.grid(True)
plt.legend()

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('placement.csv')

# Remove unwanted column
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)

# X and Y
X = df['cgpa'].values
Y = df['iq'].values

# Regression through origin
slope = np.sum(X * Y) / np.sum(X * X)

# Line starting from origin
X_line = np.linspace(0, max(X) + 1, 100)
Y_line = slope * X_line

# Plot
plt.figure(figsize=(10,6))

plt.scatter(X, Y, color='blue', s=60, label='Dataset')

plt.plot(X_line,
         Y_line,
         color='red',
         linewidth=3,
         label='Regression Line (Origin)')

plt.xlim(0, 10)
plt.ylim(0, max(Y) + 20)

plt.xlabel('CGPA')
plt.ylabel('IQ')
plt.title('Linear Regression Line Starting from Origin')
plt.grid(True)
plt.legend()

plt.show()

print("Slope =", slope)
```
