```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
af = pd.read_csv("spotifytrending_artist_byyear.csv")

# Create Copy
new_af = af.copy()


# Display Random Records
print("\n1. Random Records")
display(new_af.sample(5))

# First 5 Records
print("\n2. First 5 Records")
display(new_af.head())

# Last 5 Records
print("\n3. Last 5 Records")
display(new_af.tail())

# Number of Rows and Columns
print("\n4. Number of Rows =", new_af.shape[0])
print("Number of Columns =", new_af.shape[1])

# Data Types
print("\n5. Data Types")
display(new_af.dtypes)

# Missing Values
print("\n6. Missing Values")
display(new_af.isnull().sum())

# Columns Having Missing Values
print("\n7. Columns Having Missing Values")
display(new_af.columns[new_af.isnull().any()])

# Total Missing Values
print("\n8. Total Missing Values")
print(new_af.isnull().sum().sum())

# Mean
print("\n9. Mean")
display(new_af.mean(numeric_only=True))

# Median
print("\n10. Median")
display(new_af.median(numeric_only=True))

# Standard Deviation
print("\n11. Standard Deviation")
display(new_af.std(numeric_only=True))

# Minimum Values
print("\n12. Minimum Values")
display(new_af.min(numeric_only=True))

# Maximum Values
print("\n13. Maximum Values")
display(new_af.max(numeric_only=True))

# Most Frequent Value (Mode)
print("\n14. Most Frequent Value")
display(new_af.mode().head(1))

# Variance
variance = new_af.var(numeric_only=True)

print("\n15. Variance")
display(variance)

# Highest Variation
print("\n16. Column with Highest Variation")
print(variance.idxmax())

# Features with Greater Variance
print("\n17. Features Sorted by Variance")
display(variance.sort_values(ascending=False))

# Smallest and Largest Value
print("\n18. Smallest Value")
print(new_af.min(numeric_only=True).min())

print("\n19. Largest Value")
print(new_af.max(numeric_only=True).max())

# Valid Records
print("\n20. Valid Records")
print(new_af.dropna().shape[0])

# Samples in Each Class
print("\n21. Samples in Each Class")
display(new_af.iloc[:,0].value_counts())

# Correlation
print("\n22. Correlation Matrix")
corr = new_af.corr(numeric_only=True)
display(corr)

# Strongly Correlated Features
corr_pair = corr.unstack()
corr_pair = corr_pair[corr_pair != 1]

print("\n23. Strongly Correlated Features")
print(corr_pair.abs().sort_values(ascending=False).head(1))

# Duplicate Rows
print("\n24. Duplicate Rows")
print(new_af.duplicated().sum())

# Duplicate Columns
print("\n25. Duplicate Columns")
print(new_af.T.duplicated().sum())

# Summary Report
print("\n26. Summary Report")
display(new_af.describe(include='all'))

# -------------------------------
# HISTOGRAM
# -------------------------------
new_af.select_dtypes(include='number').hist(figsize=(10,8))
plt.suptitle("Histogram")
plt.show()

# -------------------------------
# BAR PLOT
# -------------------------------
new_af.iloc[:,0].value_counts().head(10).plot(
    kind='bar',
    figsize=(8,5),
    title="Bar Plot"
)
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()

# -------------------------------
# PIE CHART
# -------------------------------
new_af.iloc[:,0].value_counts().head(5).plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6),
    title="Pie Chart"
)
plt.ylabel("")
plt.show()


```
