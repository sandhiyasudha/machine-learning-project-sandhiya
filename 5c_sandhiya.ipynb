```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering

# Load dataset
data = pd.read_csv("student_placement_career_success_dataset.csv")

# Select career-related features
X = data[['cgpa',
          'DSA_problems_solved',
          'development_projects_count',
          'internships_completed',
          'communication_skills',
          'aptitude_score']]

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Hierarchical Clustering
model = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

# Fit the model and obtain cluster labels
cluster_labels = model.fit_predict(X_scaled)

# Display number of students in each cluster
print("Number of students in each cluster:")

for i in range(3):
    print(
        "Cluster", i + 1, ":",
        (cluster_labels == i).sum()
    )

# Plot Hierarchical Clustering results
plt.figure(figsize=(10, 6))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 5],
    c=cluster_labels,
    cmap='viridis',
    marker='o',
    s=20
)

plt.title('Hierarchical Clustering')
plt.xlabel('CGPA')
plt.ylabel('Aptitude Score')
plt.show()
```
